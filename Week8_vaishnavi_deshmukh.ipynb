**1.Import Library**

```python
import json
import re
```

**2.Calculator Tool (Baseline Math Evaluation System)**

```python
def calculator(expression):
    """
    Evaluates a mathematical expression.
    """

    try:
        result = eval(expression)
        return result
    except Exception:
        raise ValueError("Invalid mathematical expression.")
```

**3.Keyword Extraction Tool (Baseline Keyword Framework)**

```python
def extract_keywords(text):
    """
    Extracts unique keywords from text.
    """

    stop_words = {
        "the","is","a","an","and","or","to","of",
        "in","on","for","with","this","that","it"
    }

    words = re.findall(r'\b[a-zA-Z]+\b', text.lower())

    keywords = []

    for word in words:
        if word not in stop_words and word not in keywords:
            keywords.append(word)

    return keywords
```

**4.JSON Response Helper**

```python
def make_response(response_type, result):

    return {
        "type": response_type,
        "result": result
    }
```

**5.Build the Agent (Main Task Routing Pipeline)**

```python
def agent(query):

    try:

        query_lower = query.lower()

        # Route 1 → Calculator
        if "calculate" in query_lower:

            expression = query_lower.replace("calculate", "").strip()

            answer = calculator(expression)

            return make_response(
                "calculation",
                answer
            )

        # Route 2 → Keyword Extraction
        elif "keywords" in query_lower:

            text = query_lower.replace("keywords", "").strip()

            words = extract_keywords(text)

            return make_response(
                "keywords",
                words
            )

        # Route 3 → General Query
        else:

            return make_response(
                "general",
                "This query does not require a tool. General response generated."
            )

    except Exception as e:

        return make_response(
            "error",
            str(e)
        )
```

**6.Automated Validation Checks**

```python
test_queries = [

    "calculate 15 + 25",

    "keywords Artificial Intelligence is transforming healthcare and education",

    "Hello, how are you?",

    "calculate 5 / 0",

    "calculate 8 * (3 + 2)",

    "keywords Python is easy to learn and Python is powerful"
]

print("Running Validation Checks...\n")

for q in test_queries:

    print("Query :", q)

    print(json.dumps(agent(q), indent=4))

    print("-"*50)
```

**7.Interactive Agent**

```python
print("Single Agent Task Routing System")

print("Type 'exit' to stop.\n")

while True:

    query = input("Enter Query : ")

    if query.lower() == "exit":
        print("Agent Stopped.")
        break

    response = agent(query)

    print(json.dumps(response, indent=4))
```
